```python
import os
from google.colab import userdata

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")

print("API Key Connected Successfully!")
```

```python

import time
from google import genai
import os

# Connect to Gemini API
client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)

# Select Gemini model
model = "gemini-3.5-flash-lite"

# Get question from user
question = input("Ask your question: ")

# Generate AI response with retry mechanism
response = None
latency = 0

for attempt in range(1, 5):
    try:
        start = time.time()

        response = client.models.generate_content(
            model=model,
            contents=question
        )

        latency = (time.time() - start) * 1000
        break

    except Exception as e:
        print(f"Attempt {attempt} failed: {e}. Retrying...")

        if attempt < 4:
            time.sleep(attempt * 2)

# Display results
if response is not None:

    usage = response.usage_metadata

    print("\n========== GEMINI AI RESULTS ==========")

    print("\nQuestion:", question)
    print("\nAnswer:", response.text)

    print("\n========== TOKEN USAGE ==========")

    print("Input tokens:", usage.prompt_token_count)
    print("Output tokens:", usage.candidates_token_count)
    print("Total tokens:", usage.total_token_count)

    print("\nLatency (ms):", round(latency))

else:
    print("Failed to generate response after 4 attempts.")
```
